Imagine que está diseñando el primer bloque de una arquitectura convolucional para procesar imágenes satelitales multiespectrales de entrada con dimensiones $W = 256$, $H = 256$, $C = 4$ (RGB + Infrarrojo). - **Opción B (Estilo VGG):** Tres capas convolucionales secuenciales con filtros de $3 \times 3$, stride de $1$ y padding de $1$, seguidas de una capa de Max-Pooling de $2 \times 2$ (stride $2$).

### Inciso 2

Calcula matemáticamente cuántos parámetros entrenables (pesos) requiere la **Opción A** vs. las capas convolucionales de la **Opción B**. Basado en la teoría del **Campo Receptivo**, justifica por qué la industria estandarizó la **Opción B**.

#### Datos

En una convolución con:

- kernel $K \times K$
- canales de entrada $C_{in}$
- canales de salida $C_{out}$

Los **pesos** son:

$$
\#W = K^2 , C_{in}, C_{out}
$$


- Entrada: $C_{in}=4$
- Salida esperada: $C_{out}=64$

#### Opción A (AlexNet) - CONV $7\times7$

$$
\#W_A = 7^2 \cdot 4 \cdot 64
$$

$$
\#W_A = 49 \cdot 4 \cdot 64
$$

$$
\#W_A = 196 \cdot 64
$$

$$
\#W_A = 12{,}544
$$

**Requiere** $12{,}544$ **pesos**.

#### Opción B (VGG) - 3 CONV $3\times3$

Como el enunciado dice que el bloque “busca extraer $64$ canales de salida”:

- CONV1: $3\times 3$, $4 \rightarrow 64$
- CONV2: $3\times 3$, $64 \rightarrow 64$
- CONV3: $3\times 3$, $64 \rightarrow 64$

**CONV1**

$$
\#W_{B1}=3^2 \cdot 4 \cdot 64 = 9\cdot 4\cdot 64
$$

$$
\#W_{B1}=36\cdot 64=2304
$$

**CONV2**

$$
\#W_{B2}=3^2 \cdot 64 \cdot 64=9\cdot 64\cdot 64
$$

$$
\#W_{B2}=9\cdot 4096=36864
$$

**CONV3**

Es igual a CONV2:

$$
\#W_{B3}=36864
$$

## Total

$$
\#W_B = 2304 + 36864 + 36864 = 76032
$$

**Requiere** $76{,}032$ **pesos** (solo en las capas CONV).

#### Comparación

$$
\#W_A = 12{,}544
\qquad\text{vs}\qquad
\#W_B = 76{,}032
$$

#### Justificación de estandarización

La industria estandariza el patrón tipo VGG (apilar kernels pequeños) por estas razones:

**Mismo campo receptivo efectivo, más no-linealidad**

Apilar $3$ capas $3\times3$ aumenta el campo receptivo efectivo hasta $\approx 7\times7$ pero con diferencias:

- Una sola conv $7\times7$ es una transformación lineal (antes de la activación).
- Tres conv $3\times3$ tienen ReLU entre medias, o sea más “doblez” no lineal.

Resultado: más poder expresivo con la misma escala espacial de contexto.

**Mejor control del sizing**

Al insirtir en los parámetros:

- CONV pequeñas ($3\times3$)
- $S=1$
- $P=1$ para preservar tamaño y que el downsampling lo haga POOL ($2\times2$, $S=2$)

Eso hace que el diseño sea modular y evita “sizing headaches” (Opción B).

**En redes grandes, el ahorro aparece donde importa**

Aunque en este bloque aislado B puede tener más parámetros, en arquitecturas profundas:

- Kernels grandes al inicio son costosos, y la estrategia de bloques $3\times3$ se vuelve una regla sistemática y eficiente, además de que el verdadero monstruo de parámetros históricamente han sido las **FC**.


#### Conclusión

- Opción B se estandariza porque da campo receptivo grande de forma progresiva, con más no-linealidades, y con un patrón modular que mantiene dimensiones estables y hace el diseño escalable.
- La comparación de parámetros depende del ancho de canales por capa. En tu configuración concreta, B tiene más pesos porque sostienes $64$ canales en 3 capas.

**Respuesta:**

Los métodos tradicionales **SIFT** y **HOG** pertenecen al paradigma de extracción manual de características, donde primero se detectan puntos o gradientes y luego se construyen descriptores numéricos que se entregan a un clasificador como **SVM**. Este enfoque funciona bien cuando las estructuras visuales son relativamente estables, pero en entornos agrícolas reales las hojas presentan **variaciones de iluminación, rotación, escala, perspectiva y oclusiones parciales** (características locales). Bajo estas condiciones, los descriptores manuales capturan patrones rígidos basados en gradientes o texturas locales que pueden cambiar significativamente dependiendo del ángulo de captura, la hora del día o el estado fisiológico de la planta. Como resultado, el sistema tiende a perder consistencia cuando intenta reconocer síntomas de enfermedad que no aparecen siempre con la misma forma o intensidad.

Las **redes neuronales convolucionales profundas** aprenden automáticamente representaciones visuales jerárquicas mediante **mapas de características**. Las primeras capas aprenden filtros simples que detectan bordes y contrastes, las capas intermedias capturan texturas y patrones visuales más complejos, y las capas profundas integran estas señales para formar representaciones semánticas de objetos o estructuras biológicas. Este aprendizaje jerárquico permite que la red reconozca una enfermedad incluso cuando la hoja cambia de orientación, iluminación o contexto visual. Además, mediante **Transfer Learning** es posible reutilizar redes previamente entrenadas en grandes bases de datos visuales y adaptarlas al cultivo de mango, lo que permite obtener modelos robustos sin necesidad de diseñar manualmente descriptores específicos para cada variación visual del entorno agrícola.

**Respuesta:**

Elegiría **VGG-16** porque su arquitectura modular basada en bloques repetidos de convoluciones $3 \times 3$ permite construir representaciones visuales de manera progresiva y sistemática. A diferencia de AlexNet, que utiliza filtros de distintos tamaños de forma más empírica (por ejemplo $11 \times 11$, $5 \times 5$, $3 \times 3$), VGG mantiene un diseño homogéneo que facilita la extracción jerárquica de patrones visuales. Esta modularidad permite que la red capture estructuras visuales de forma gradual, desde bordes y texturas hasta patrones complejos. Esto significa que VGG puede construir representaciones más ricas y semánticas de las hojas, identificando patrones sutiles de daño o infección incluso bajo variaciones visuales del entorno agrícola.
```python
import os, time, copy, random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim

from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader, random_split, Subset
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt

# Reproducibilidad
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

set_seed(42)

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# Parámetros
IMG_SIZE = 64
BATCH_SIZE = 64 if torch.cuda.is_available() else 32
NUM_WORKERS = 0 
PIN_MEMORY = True if torch.cuda.is_available() else False

data_dir = "./archive"

# Normalización ImageNet
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

data_transforms = {
    "train": transforms.Compose([
        transforms.RandomResizedCrop(IMG_SIZE),
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(10),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ]),
    "val": transforms.Compose([
        transforms.Resize(int(IMG_SIZE * 1.15)),
        transforms.CenterCrop(IMG_SIZE),
        transforms.ToTensor(),
        transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    ])
}

# Dataset
base_ds = datasets.ImageFolder(data_dir)

class_names = base_ds.classes
num_classes = len(class_names)

print("Clases:", num_classes, class_names)
print("Total imágenes:", len(base_ds))

# Split train / val
train_size = int(0.8 * len(base_ds))
val_size = len(base_ds) - train_size

train_subset, val_subset = random_split(base_ds, [train_size, val_size])

# Wrapper para transforms
class SubsetWithTransform(torch.utils.data.Dataset):
    def __init__(self, subset, transform=None):
        self.subset = subset
        self.transform = transform

    def __len__(self):
        return len(self.subset)

    def __getitem__(self, idx):
        x, y = self.subset[idx]
        if self.transform:
            x = self.transform(x)
        return x, y

train_ds = SubsetWithTransform(train_subset, transform=data_transforms["train"])
val_ds   = SubsetWithTransform(val_subset, transform=data_transforms["val"])

# Reducir dataset (para velocidad)
MAX_TRAIN_SAMPLES = 800
MAX_VAL_SAMPLES = 200

if len(train_ds) > MAX_TRAIN_SAMPLES:
    train_ds = Subset(train_ds, range(MAX_TRAIN_SAMPLES))

if len(val_ds) > MAX_VAL_SAMPLES:
    val_ds = Subset(val_ds, range(MAX_VAL_SAMPLES))

print("Train samples:", len(train_ds))
print("Val samples:", len(val_ds))

# DataLoaders
dataloaders = {
    "train": DataLoader(train_ds,
                        batch_size=BATCH_SIZE,
                        shuffle=True,
                        num_workers=NUM_WORKERS,
                        pin_memory=PIN_MEMORY),

    "val": DataLoader(val_ds,
                      batch_size=BATCH_SIZE,
                      shuffle=False,
                      num_workers=NUM_WORKERS,
                      pin_memory=PIN_MEMORY)
}

# Modelo VGG16
model = models.vgg16(weights=models.VGG16_Weights.IMAGENET1K_V1)

# Freeze convolucionales
for p in model.features.parameters():
    p.requires_grad = False

# Nuevo classifier
model.classifier = nn.Sequential(
    nn.Linear(25088, 1024),
    nn.ReLU(),
    nn.Dropout(0.5),
    nn.Linear(1024, 256),
    nn.ReLU(),
    nn.Dropout(0.3),
    nn.Linear(256, num_classes)
)

model = model.to(device)

print(model.classifier)

# Loss y optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.classifier.parameters(), lr=1e-3)

# Training loop
def train_model(model, dataloaders, num_epochs=5):

    for epoch in range(num_epochs):

        print(f"\nEpoch {epoch+1}/{num_epochs}")
        print("-"*25)

        for phase in ["train", "val"]:

            if phase == "train":
                model.train()
            else:
                model.eval()

            running_loss = 0
            running_corrects = 0
            total_samples = 0

            for inputs, labels in dataloaders[phase]:

                inputs = inputs.to(device)
                labels = labels.to(device)

                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == "train"):

                    outputs = model(inputs)
                    loss = criterion(outputs, labels)

                    _, preds = torch.max(outputs, 1)

                    if phase == "train":
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels).item()
                total_samples += inputs.size(0)

            epoch_loss = running_loss / total_samples
            epoch_acc = running_corrects / total_samples

            print(f"{phase} Loss: {epoch_loss:.4f} | Acc: {epoch_acc:.4f}")

    return model

# -----------------------
# Entrenar
# -----------------------
model = train_model(model, dataloaders, num_epochs=5)
```

### 4. Evaluación y Reporte


```python
def evaluate_and_confusion(model, dataloader, class_names):
    model.eval()
    all_preds, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device, non_blocking=True)
            labels = labels.to(device, non_blocking=True)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            all_preds.extend(preds.cpu().numpy().tolist())
            all_labels.extend(labels.cpu().numpy().tolist())

    cm = confusion_matrix(all_labels, all_preds)

    # accuracy global
    acc = (np.array(all_preds) == np.array(all_labels)).mean()

    # Plot matriz de confusión (sin seaborn)
    plt.figure(figsize=(10, 8))
    plt.imshow(cm, interpolation="nearest")
    plt.title("Matriz de Confusión (Validación)")
    plt.colorbar()
    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names, rotation=45, ha="right")
    plt.yticks(ticks, class_names)
    plt.ylabel("Etiqueta real")
    plt.xlabel("Predicción")

    thresh = cm.max()/2.0 if cm.max() else 0.5
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, str(cm[i, j]),
                     ha="center", va="center",
                     color="white" if cm[i, j] > thresh else "black")

    plt.tight_layout()
    plt.show()

    return acc, cm

val_acc, cm = evaluate_and_confusion(model, dataloaders["val"], class_names)
print("Accuracy validación:", val_acc)
```

### Reporte Final

**Precisión Global:**  
Después de aplicar Transfer Learning con VGG-16 preentrenado en ImageNet y entrenar únicamente el clasificador final durante 5 épocas, el modelo obtuvo una **precisión de validación de 0.845 (84.5%)**.  
Este resultado indica que los filtros previamente aprendidos por la red para detectar bordes, texturas y patrones generales fueron efectivos para diferenciar varias de las enfermedades presentes en las hojas de mango.

**Análisis de la Matriz de Confusión:**  
La matriz de confusión muestra que varias clases como **Bacterial Canker, Cutting Weevil y Die Back** fueron clasificadas correctamente en la mayoría de los casos.

**Conclusión sobre Sobreajuste (Overfitting):**  
No se observan indicios fuertes de sobreajuste, ya que el modelo generaliza razonablemente bien sobre el conjunto de validación.

No obstante, entrenar durante más épocas o utilizar un conjunto de entrenamiento más grande podría mejorar aún más la precisión del sistema.
